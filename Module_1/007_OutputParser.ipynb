# Output Parsers

Output parsers transform the llm's raw text output into structured python objects
| Parser Name | Output | Output Type |
|-------------|--------|-------------|
| StrOutputParser | Plain string output (clean text from LLM) | string |
| JsonOutputParser | Valid JSON / dictionary format | dict / JSON |
| PydanticOutputParser | Python Pydantic model object | Python object (Pydantic model) |
| CommaSeparatedListOutputParser | Python list from comma-separated text | list[str] |
| StructuredOutputParser | Schema-based structured JSON output | dict / JSON |
| RegexParser | Extracted value based on regex pattern | string / matched groups |
| OutputFixingParser | Automatically fixed valid JSON output | dict / JSON |
| RetryOutputParser | Re-generates output if parsing fails | same as target parser output |

```python
# Import Gemini-ChatGoogleGenerativeAI 
from langchain_google_genai import ChatGoogleGenerativeAI
import os
from dotenv import load_dotenv
load_dotenv()

GOOGLE_API_KEY = os.getenv("GEMINI_API_KEY")
print(GOOGLE_API_KEY)

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash", 
    api_key= GOOGLE_API_KEY
    
    )
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser


json_prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Always response with valid JSON."),
        ("human", "Give me info about {person} as JSON with keys: name, field , acheivement")
    ]
)

chain= json_prompt | llm | JsonOutputParser()

result= chain.invoke({"person": "Mark Zuckerberg"})
print(result)
print(result['name'])
print(result['field'])
print(result['achievement'])
```
