# 📦 Output Parsers — LangChain Bangla

LLM সবসময় plain text রিটার্ন করে।  
কিন্তু আমাদের প্রায়ই **structured data** দরকার হয় — যেমন JSON, Python object, list ইত্যাদি।

**Output Parser** হলো সেই টুল যা LLM-এর raw text কে কাজের মতো structured ফরম্যাটে রূপান্তর করে।

---
## Output Parser-এর প্রকারভেদ:
| Parser | আউটপুট ধরন | কখন ব্যবহার করবেন |
|--------|------------|-------------------|
| `StrOutputParser` | সাধারণ string | সহজ টেক্সট উত্তরের জন্য |
| `JsonOutputParser` | dict / JSON | structured data দরকার হলে |
| `PydanticOutputParser` | Python object | type-safe data দরকার হলে |
| `CommaSeparatedListOutputParser` | list | আইটেমের তালিকার জন্য |

---
## এই নোটবুকে যা দেখব:

`JsonOutputParser` ব্যবহার করে মার্ক জাকারবার্গের তথ্য structured JSON-এ বের করব।

**Chain:** `prompt | llm | JsonOutputParser()`


```python
# Import Gemini-ChatGoogleGenerativeAI 
from langchain_google_genai import ChatGoogleGenerativeAI
import os
from dotenv import load_dotenv
load_dotenv()

GOOGLE_API_KEY = os.getenv("GEMINI_API_KEY")
print(GOOGLE_API_KEY)

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash", 
    api_key= GOOGLE_API_KEY
    
    )
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser


json_prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Always response with valid JSON."),
        ("human", "Give me info about {person} as JSON with keys: name, field , acheivement")
    ]
)

chain= json_prompt | llm | JsonOutputParser()

result= chain.invoke({"person": "Mark Zuckerberg"})
print(result)
print(result['name'])
print(result['field'])
print(result['achievement'])
```
