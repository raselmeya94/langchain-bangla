### Load API KEY

```python
import os
from dotenv import load_dotenv
load_dotenv()

GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
# print(GOOGLE_API_KEY)
```

```python
# Import Gemini-ChatGoogleGenerativeAI 
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate #003_prompt_engineering
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash", 
    api_key= GOOGLE_API_KEY
    
    )

# ChatPromptTemplate - the modern, recommended approach

chat_template= ChatPromptTemplate.from_messages(
    [
        ("system", "Your are an expert of {Greeting} with polite and clam."),
        ("human", "My name is {Name}")
    ]
)

messages= chat_template.format_messages(
    Greeting="Greeting",
    Name="Rasel"

)

for msg in messages:
    print(f"{type(msg).__name__} : {msg.content}")

# llm invoke for Generate 
response= llm.invoke(messages)
```

### Raw Response

```python
from pprint import pprint

pprint(response.__dict__)
```

### Parse Actual Response

```python
print(response.content)
```

### FewShow QA

```python
from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate

# FewShotPromptTemplate - tech the model with examples

examples= [
    {"word": "Happy", "antonym": "Sad"},
    {"word": "Tall", "antonym": "Short"},
    {"word": "Fast", "antonym": "Slow"},
]

example_template= PromptTemplate(
    input_variables=['word', 'antonym'],
    template= "word: {word}\nAntonym: {antonym}"
)

few_shot= FewShotPromptTemplate(
    examples= examples,
    example_prompt= example_template,
    prefix= "Given the antonym of each word:",
    suffix= "Word: {input}\nAntonym:",
    input_variables=['input']
)

few_shot_formatted= few_shot.format(
    input="Black"
)

llm.invoke(few_shot_formatted).content
```
