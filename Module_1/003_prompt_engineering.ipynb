## Prompt Template

```python
from langchain_core.prompts import PromptTemplate

#Basic PromptTemplate - use  { variable } placeholders

template= PromptTemplate(
    input_vairable=['Name', "Greeting"],
    template= "{Greeting} to new session, {Name}."
)


# Format the prompt

formatted= template.format(Name="Rasel", Greeting="Welcome")
print(formatted)
```

## ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate

# ChatPromptTemplate - the modern, recommended approach

chat_template= ChatPromptTemplate.from_messages(
    [
        ("system", "Your are an expert of {Greeting} with polite and clam."),
        ("human", "My name is {Name}")
    ]
)


messages= chat_template.format_messages(
    Greeting="Greeting",
    Name="Rasel"

)

print(messages)
```

```python
for msg in messages:
    print(f"{type(msg).__name__} : {msg.content}")
```

## FewShotPromptTemplate

```python
from langchain_core.prompts import FewShotPromptTemplate

# FewShotPromptTemplate - tech the model with examples

examples= [
    {"word": "Happy", "antonym": "Sad"},
    {"word": "Tall", "antonym": "Short"},
    {"word": "Fast", "antonym": "Slow"},
]

example_template= PromptTemplate(
    input_variables=['word', 'antonym'],
    template= "word: {word}\nAntonym: {antonym}"
)

few_shot= FewShotPromptTemplate(
    examples= examples,
    example_prompt= example_template,
    prefix= "Given the antonym of each word:",
    suffix= "Word: {input}\nAntonym:",
    input_variables=['input']
)

formatted= few_shot.format(
    input="Black"
)
```

```python
print(formatted)
```
