# 💬 LangChain দিয়ে কথোপকথনের ইতিহাস রাখা — LangChain Bangla

সাধারণ LLM call-এ প্রতিটি প্রশ্ন আলাদা — মডেল আগের কথা মনে রাখে না।  
কিন্তু chatbot তৈরি করতে হলে আগের কথোপকথন মনে রাখা দরকার।

## সমাধান: `MessagesPlaceholder`

`MessagesPlaceholder` হলো একটি বিশেষ placeholder যেখানে পুরো conversation history ঢোকানো যায়।  
ফলে মডেল আগের সব কথা দেখে উত্তর দিতে পারে।

---
## এই নোটবুকে যা শিখব:
| বিষয় | বিবরণ |
|-------|-------|
| `MessagesPlaceholder` | History ইনজেক্ট করার পদ্ধতি |
| `HumanMessage` / `AIMessage` | মেসেজ তৈরির ক্লাস |
| Manual history inject | Python list-এ history রেখে pass করা |

---


````python
# Import Gemini-ChatGoogleGenerativeAI 
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate #003_prompt_engineering
import os
from dotenv import load_dotenv
load_dotenv()

GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
# print(GOOGLE_API_KEY)

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash", 
    api_key= GOOGLE_API_KEY
    
    )
````

## 🔧 কীভাবে কাজ করে?

```
System Message
    ↓
history = [HumanMessage, AIMessage, HumanMessage, AIMessage ...]
    ↓
নতুন Human Message
    ↓
LLM উত্তর দেয়
```

`history` list-এ আগের সব কথোপকথন রাখা হয়।  
প্রতিবার নতুন প্রশ্নের সাথে পুরো history পাঠানো হয় — তাই মডেল আগের কথা "মনে রাখতে" পারে।


````python
# MessagesPlaceholder= Dynamic injection of conversation history

from langchain_core.prompts import MessagesPlaceholder, ChatPromptTemplate
from langchain_core.messages import HumanMessage, AIMessage
chat_with_history= ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant."),
        MessagesPlaceholder(variable_name="history"),
        ("human", "{input_question}")
    ]
)

history= [
    HumanMessage(content="My name is Rasel"),
    AIMessage(content="Nice to meet you, Rasel!"),
]

messages= chat_with_history.format_messages(history=history, input_question="What is my Name?")

````

````python
llm.invoke(messages).content
````
